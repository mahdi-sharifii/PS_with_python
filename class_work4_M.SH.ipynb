```python
#mehdi_sharifi
import math as m
```

```python
def y1(x):
    return m.sqrt(m.sin(x)+1)
```

```python
y1(4)
```

```python
def y2(x):
    return m.sqrt(abs(m.sin(x)/m.cos(x)))
```

```python
y2(5)
```

```python
def y3(x):
    return m.sqrt(m.tan(x)/m.sin(x))
    
```

```python
y3(5)
```

```python
def y4(x, e):
    return m.sqrt(1/(1+(e**-x)))
```

```python
y4(7,2)
```

```python
def y5(w,a):
    return w-a*(m.sqrt(w)/m.sqrt(a))
```

```python
y5(5,6)
```

```python
def y6(e,x):
    return 1-(e**m.sin(x))
```
